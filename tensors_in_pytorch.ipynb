### Mathematical Operations Supported on Integer Tensors in PyTorch

When working with **integer tensors** (such as `torch.int32`, `torch.int64`/`Long`), PyTorch supports various operations. They generally fall into three categories:

#### 1. Operations Supported Natively
These operations do not require any casting and work directly on integer types:
* **Basic Arithmetic**: Addition (`+`), Subtraction (`-`), Multiplication (`*`), Integer/Floor Division (`//`), Modulo/Remainder (`%`), and Power (`**` with non-negative integer exponents).
* **Absolute Value**: `torch.abs(tensor)`
* **Negation**: `torch.neg(tensor)` or `-tensor`
* **Clamping/Clipping**: `torch.clamp(tensor, min, max)`
* **Basic Reductions**:
  * Summation: `torch.sum(tensor)`
  * Product: `torch.prod(tensor)`
  * Minimum: `torch.min(tensor)`
  * Maximum: `torch.max(tensor)`
  * Median: `torch.median(tensor)`
* **Comparison Operations**: `<`, `>`, `<=`, `>=`, `==`, `!=` (returns boolean tensors).
* **Bitwise Operations**: Bitwise AND (`&`), OR (`|`), XOR (`^`), NOT (`~`), and Bit shifts (`<<`, `>>`).

#### 2. Operations Requiring Float Casts (Not supported natively on integers)
These functions mathematically require continuous/fractional outputs and will raise a `RuntimeError` if called directly on integer tensors:
* **Statistical Operations**: Mean (`torch.mean()`), Standard Deviation (`torch.std()`), and Variance (`torch.var()`).
* **Rounding & Truncation**: Ceil (`torch.ceil()`), Floor (`torch.floor()`), Round (`torch.round()`), and Trunc (`torch.trunc()`). (Note: Rounding functions are redundant on integers, but still require float inputs in PyTorch).
* **Trigonometric & Exponential**: `torch.sin()`, `torch.cos()`, `torch.exp()`, `torch.log()`, `torch.sqrt()`, etc.
* **True Division**: The true division operator (`/`) on integer tensors implicitly converts the result to a float tensor.

*To use any of these unsupported operations, cast your integer tensor to a float first using `.float()` or `.to(torch.float32)`.*
